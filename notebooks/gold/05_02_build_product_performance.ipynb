# Build Product Performance
Aggregates Silver Layer data to create product-level sales and performance metrics.

```sql
CREATE OR REPLACE TABLE ecommerce_catalog.gold.product_performance AS
SELECT 
    p.product_id,
    p.product_name,
    p.category,
    COUNT(DISTINCT o.order_id)      AS order_count,
    COALESCE(SUM(o.quantity), 0)    AS units_sold,
    ROUND(COALESCE(SUM(o.quantity * o.unit_price), 0.0),2) AS revenue,
    ROUND(
        COALESCE(
            SUM(o.quantity * o.unit_price) / SUM(o.quantity), 0.0
        ), 2) AS avg_selling_price

FROM ecommerce_catalog.silver.silver_products p
LEFT JOIN ecommerce_catalog.silver.silver_orders o 
    ON p.product_id = o.product_id
GROUP BY 
    p.product_id,
    p.product_name,
    p.category
ORDER BY 
    revenue DESC;
```
