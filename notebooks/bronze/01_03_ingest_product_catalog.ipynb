# Ingest Product Catalog
Ingests raw product catalog data into the Bronze Layer for further validation and transformation.

```python
from pyspark.sql.functions import current_timestamp
from delta.tables import DeltaTable
```

```python
file_path = "/Volumes/ecommerce_catalog/raw/landing_zone/product_catalog_2026_01.json"

df = spark.read.option("multiLine", True).json(file_path)
    

display(df)
```

```python
df = df.withColumn("process_date", current_timestamp())
display(df)
```

```python
df = df.select("*", "_metadata.file_name").withColumnRenamed("file_name", "source_file")
display(df)
```

```python
df.write.format("delta")\
    .mode("append")\
    .saveAsTable("ecommerce_catalog.bronze.product_catalog_raw")
```
