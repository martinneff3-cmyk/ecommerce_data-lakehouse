# Ingest Customers
Ingests raw customer data into the Bronze Layer for further validation and transformation.

```python
from pyspark.sql import functions as F
```

```python
file_path = "/Volumes/ecommerce_catalog/raw/landing_zone/customer_2026_01.csv"

(
    spark.read.format("csv") 
    .option("header", "true") 
    .option("inferSchema", "true") 
    .load(file_path)
    .withColumn("process_date", F.current_timestamp())
    .select(
        "*", 
        F.col("_metadata.file_name").alias("source_file"))
    .write.format("delta")
    .mode("overwrite")
    .saveAsTable("ecommerce_catalog.bronze.customers_raw")
)

```
