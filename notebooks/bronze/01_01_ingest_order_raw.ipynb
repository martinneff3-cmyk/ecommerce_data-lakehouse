# Ingest Orders
Ingests raw Order data into the Bronze Layer for further validation and transformation.

```python
from pyspark.sql.functions import current_timestamp
from delta.tables import DeltaTable
```

```python
file_path = "/Volumes/ecommerce_catalog/raw/landing_zone/orders_2026_01.csv"

df = spark.read.format("csv") \
    .option("header", "true") \
    .option("inferSchema", "true") \
    .load("dbfs:/Volumes/ecommerce_catalog/raw/landing_zone/orders_2026_01.csv")


display(df)
```

```python
df = df.withColumn("processDate", current_timestamp())
display(df)
    
```

```python
df = df.select("*", "_metadata.file_name").withColumnRenamed("file_name", "source_file")
display(df)
```

```python
df = df.withColumnRenamed("processDate", "process_date")
display(df)
```

```python
df.write.format("delta")\
    .mode("append")\
    .saveAsTable("ecommerce_catalog.bronze.orders_raw")
```
