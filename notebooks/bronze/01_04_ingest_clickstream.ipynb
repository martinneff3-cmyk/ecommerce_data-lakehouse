# Ingest Clickstream
Ingests raw clickstream data into the Bronze Layer for further validation and transformation.

```python
from pyspark.sql import functions as F
```

```python
file_path = "/Volumes/ecommerce_catalog/raw/landing_zone/clickstream_2026_01.json"

(
    spark.read
    .format("json")
    .option("multiLine", True)
    .load(file_path)
    .withColumn("process_date", F.current_timestamp())
    .select(
        "*", 
        F.col("_metadata.file_name").alias("source_file"))
    .write
    .mode("overwrite")
    .saveAsTable("ecommerce_catalog.bronze.clickstream_raw")
    
)
```
