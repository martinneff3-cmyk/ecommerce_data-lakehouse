```sql
CREATE CATALOG IF NOT EXISTS ecommerce_catalog
COMMENT 'Catalog for the E-Commerce Data Lakehouse portfolio project';

CREATE SCHEMA IF NOT EXISTS ecommerce_catalog.bronze
COMMENT 'Raw ingested data stored as Delta tables';

CREATE SCHEMA IF NOT EXISTS ecommerce_catalog.silver
COMMENT 'Cleaned, standardized and quality_checked data';

CREATE SCHEMA IF NOT EXISTS ecommerce_catalog.gold
COMMENT 'Business-ready analytics tables and KPIs';

CREATE SCHEMA IF NOT EXISTS ecommerce_catalog.raw
COMMENT 'Landing area for raw source files';

CREATE VOLUME IF NOT EXISTS ecommerce_catalog.raw.landing_zone
COMMENT 'Managed volume for raw CSV, JSON and simulated source files';
```
